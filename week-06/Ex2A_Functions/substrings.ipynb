## Lab 1. Working with names – parsing strings

```python
name = 'faiz khan'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
```

```python
parse_and_display(name)
```

```python
names = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(names[0])
```

```python
parse_and_display(names[1])
```

```python
parse_and_display('Faiz Khan')
```
