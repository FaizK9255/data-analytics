# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

 ## Looping Through Arrays

```python
arrA = np.array([
    [11,12,13],
    [14,15,16]
])
```

```python
print(arrA)
```

## loop through the first dimension

```python
for x in arrA:
    print(x * 2)
```

## loop through the second dimension

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

### the first dimension keeps the numbers in a list and multiplies them, while the second dimension seems to break into its own output and also multiply the numbers

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# the for loop goes through each group, repeats each item in those groups, then multiplies by 2.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)

# you get a value error that all input arrays must have the same shape
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
print(another_2)

# got ann error that states all the input array dimensions except for the concatenation must match
```
