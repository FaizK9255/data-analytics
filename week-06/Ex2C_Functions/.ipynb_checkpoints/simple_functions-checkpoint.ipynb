```python
def favorite_things(name, movie, music):
    return f'My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician/band is {music}.'
```

```python
print(favorite_things("Faiz", "Warrior", "Brent Faiyaz"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Yosemite", "I love the scenery, the escape from normal life, and snowbaording in the Winter"))
```

```python
print(favorite_place("my hometown", "I get to be with friends and family and live the lifestyle I grew up with"))
```
